### インポート


```python
# --- セル1: 必要ライブラリのインポート ---
# LangChainを利用して複数の回答を取得
# collections.Counterを利用して多数決を実装
from langchain.chat_models import ChatOpenAI
from langchain.prompts import PromptTemplate
from collections import Counter
```

### LLM の設定・プロンプトの準備


```python
# --- セル2: モデルとプロンプトの準備 ---
# ここでは同じモデルをtemperatureを上げて複数回呼び出し、
# ニューラちゃんの「春風の線」に相当する多様な回答を生成します。

llm = ChatOpenAI(model="gpt-4o", temperature=0.7)  # モデル指定
prompt = PromptTemplate.from_template("次の質問に答えてください: {question}")

question = "AIエージェントの利点を1つ挙げてください。"
```

### 複数回答の生成


```python
# --- セル3: 複数回答の生成 ---
# 同じ質問を5回繰り返して、複数のAI回答を収集します。
answers = [llm.invoke(prompt.format(question=question)).content for _ in range(5)]

print("=== 個別のAI回答（春風の線） ===")
for i, ans in enumerate(answers, 1):
    print(f"回答 {i}: {ans}")
```

### 投票（多数決）処理


```python
# --- セル4: 投票（多数決）処理 ---
# Counterを使って最も多い回答を「多数派の風」として決定します。
vote = Counter(answers)
final_answer = vote.most_common(1)[0][0]

print("\n=== 投票結果 ===")
print("多数派の風（最終回答）:", final_answer)

print("\n=== 少数派の風（逆風） ===")
for ans, count in vote.items():
    if ans != final_answer:
        print(f"- {ans} （{count}票）")
```

### まとめ


```python
# --- セル5: まとめ ---
# ニューラちゃんが示した「多数派の春風」と、
# 孫ハルシ義が守った「逆風」を出力として整理します。
print("🌸 最終まとめ 🌸")
print(f"多数派の春風: {final_answer}")
print("逆風も次の判断や改善に活かされます。")
```
