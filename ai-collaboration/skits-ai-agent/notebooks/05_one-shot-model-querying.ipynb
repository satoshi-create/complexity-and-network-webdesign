### API キーの取得


```python
from google.colab import userdata
import os

# サイドバーで追加したシークレットを取得
apikey = userdata.get("OPENAI_API_KEY")

# 改行や空白を除去して環境変数に登録
if apikey:
    os.environ["OPENAI_API_KEY"] = apikey.strip()
else:
    raise ValueError("ColabのSecretsに OPENAI_API_KEY が設定されていません")
```

### インポート


```python
# ============================================================
# セル1: 必要なライブラリのインストールとインポート
# ============================================================

# LangChainのOpenAIラッパーを利用
# pip install langchain-openai が必要です
from langchain_openai import ChatOpenAI
```

### LLM の設定


```python
# ============================================================
# セル2: モデルの準備
# ============================================================

# 軽量で低コストのモデルを選択（gpt-4o-miniなど）
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
```

### 情報不足のプロンプト（失敗例）


```python
# ============================================================
# セル3: 情報不足のプロンプト（失敗例）
# ============================================================

# 高速入口でテンパったさとが曖昧に聞く
prompt_bad = "ETCどう使うの？"

response_bad = llm.invoke(prompt_bad)

print("【失敗プロンプト】")
print(prompt_bad)
print("\n【ETC姉の回答】")
print(response_bad.content)
```

### 正しいプロンプト（成功例）


```python
# ============================================================
# セル4: 正しいプロンプト（成功例）
# ============================================================

# 情報をきちんと詰め込んだプロンプト
prompt_good = (
    "ETC初心者です。"
    "高速道路入口で使う方法を3ステップで説明してください。"
    "カードの入れ方・差し込み場所・注意点を含めて、100文字以内で簡潔に。"
)

response_good = llm.invoke(prompt_good)

print("【成功プロンプト】")
print(prompt_good)
print("\n【ETC姉の回答】")
print(response_good.content)
```
